```python
import pandas as pd

# Charger les features déjà calculées
user_volume = pd.read_csv("../data/processed/user_volume_recency.csv")
category_entropy = pd.read_csv("../data/processed/user_category_entropy.csv")
user_style = pd.read_csv("../data/processed/user_style_features.csv")
user_temporal = pd.read_csv("../data/processed/user_temporal_features.csv")

# Sécuriser user_id
for df_ in [user_volume, category_entropy, user_style, user_temporal]:
    df_["user_id"] = df_["user_id"].astype(str).str.strip()
```

```python
user_behavior = (
    user_volume
    .merge(
        category_entropy[["user_id", "category_entropy"]],
        on="user_id",
        how="left"
    )
    .merge(
        user_style[[
            "user_id",
            "avg_rating",
            "rating_std",
            "avg_review_length",
            "avg_sentiment"
        ]],
        on="user_id",
        how="left"
    )
    .merge(
        user_temporal[["user_id", "weekend_activity_ratio"]],
        on="user_id",
        how="left"
    )
)
```

```python
user_behavior_features = user_behavior[[
    "user_id",
    "nb_interactions",
    "recency_days",
    "nb_products",
    "category_entropy",
    "avg_rating",
    "rating_std",
    "avg_review_length",
    "avg_sentiment",
    "weekend_activity_ratio"
]]
```

```python
user_behavior_features = user_behavior_features.fillna(0)
```

```python
user_behavior_features.to_csv(
    "../data/processed/user_behavior_features.csv",
    index=False
)
```
