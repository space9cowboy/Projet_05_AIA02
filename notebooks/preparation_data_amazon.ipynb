```python
# Conversion du timestamp
df["timestamp"] = pd.to_datetime(df["timestamp"], unit="ms")
```

```python
#Encodage des IDs (pour le ML)
df["user_id_enc"] = df["user_id"].astype("category").cat.codes
df["item_id_enc"] = df["parent_asin"].astype("category").cat.codes
```

```python
df["rating"] = df["rating"].fillna(df["rating"].mean())
```

```python
df = df.dropna(subset=["user_id", "parent_asin"])
```

```python
# Standardisation robuste avec gestion std=0
mean = df["rating"].mean()
std = df["rating"].std()
if std == 0 or pd.isna(std):
    # si pas de variance, mettre 0 partout (ou NaN selon votre choix)
    df["rating_norm"] = 0.0
else:
    df["rating_norm"] = (df["rating"] - mean) / std

```

```python
#
df_model = df[[
    "user_id_enc",
    "item_id_enc",
    "rating_norm",
    "timestamp"
]].copy()

df_model.head()
```
